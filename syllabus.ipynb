# Capstone Project  

The goal of the capstone project is to give students the opportunity to perform a data science analysis on a real-world problem. Students will work in small groups to formulate a data science problem, collect relevant data, and follow all steps of the data science process. The capstone provides students experience collaborating in teams and results in a concrete project that they can list on their resume and discuss during interviews.  

The capstone project is set up to have multiple deliverables throughout the project to ensure that students are making sufficient progress.
